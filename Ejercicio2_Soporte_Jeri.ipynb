# Ejercicio 2: Escalamiento de Tickets de Soporte Técnico

Este ejercicio implementa un agente inteligente que decide a qué nivel de
soporte debe enviarse un ticket, considerando el tiempo de espera, el nivel
de urgencia y si el cliente es premium.

### Ficha PEAS

**Percepción (S):**
- Tiempo de espera del ticket en minutos.
- Nivel de urgencia: baja, media o alta.
- Condición del cliente: premium o no premium.

**Acciones (A):**
- Mantener el ticket en nivel 1.
- Escalar el ticket a nivel 2.
- Escalar el ticket a nivel 3.

**Entorno (E):**
- Mesa de ayuda o sistema de soporte técnico.
- Tickets generados por clientes.
- Equipo de soporte de diferentes niveles.

**Objetivo:**
Asignar cada ticket al nivel de soporte adecuado, considerando su urgencia,
tiempo de espera y condición del cliente.

**Medida de desempeño (P):**
- Tiempo de respuesta.
- Correcta priorización de tickets.
- Reducción de tiempos de espera.
- Atención adecuada de tickets urgentes.
- Consistencia en la aplicación de las reglas.
