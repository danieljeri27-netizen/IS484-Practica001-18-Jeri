# Ejercicio 2: Escalamiento de Tickets de Soporte Técnico

Este ejercicio implementa un agente inteligente que decide a qué nivel de
soporte debe enviarse un ticket, considerando el tiempo de espera, el nivel
de urgencia y si el cliente es premium.

### Ficha PEAS

**Percepción (S):**
- Tiempo de espera del ticket en minutos.
- Nivel de urgencia: baja, media o alta.
- Condición del cliente: premium o no premium.

**Acciones (A):**
- Mantener el ticket en nivel 1.
- Escalar el ticket a nivel 2.
- Escalar el ticket a nivel 3.

**Entorno (E):**
- Mesa de ayuda o sistema de soporte técnico.
- Tickets generados por clientes.
- Equipo de soporte de diferentes niveles.

**Objetivo:**
Asignar cada ticket al nivel de soporte adecuado, considerando su urgencia,
tiempo de espera y condición del cliente.

**Medida de desempeño (P):**
- Tiempo de respuesta.
- Correcta priorización de tickets.
- Reducción de tiempos de espera.
- Atención adecuada de tickets urgentes.
- Consistencia en la aplicación de las reglas.

### Reglas de decisión

Las reglas utilizadas por el agente son:

1. Si la urgencia es **alta**, el ticket se **escala a nivel 3**
   inmediatamente.

2. Si el tiempo de espera es **mayor o igual a 60 minutos**, se
   **escala a nivel 2**.

3. Si la urgencia es **media** y el tiempo de espera es **mayor o igual a
   30 minutos**, se **escala a nivel 2**.

4. Si el cliente es **premium** y el tiempo de espera es **mayor o igual a
   20 minutos**, se **escala a nivel 2**.

5. Si ninguna de las condiciones anteriores se cumple, el ticket se
   **mantiene en nivel 1**.


Las reglas buscan reducir el tiempo de atención de los tickets que requieren
una respuesta prioritaria. Una urgencia alta se escala directamente al nivel
3 porque requiere atención inmediata. Los tiempos de espera elevados y la
condición premium justifican una escalación a nivel 2 cuando alcanzan los
límites establecidos. De esta manera se busca equilibrar las necesidades del
cliente con la capacidad del equipo de soporte.

### Código

```python
def agente_soporte(
    tiempo_espera_minutos,
    nivel_urgencia,
    cliente_premium
):
    
    if nivel_urgencia == "alta":
        return (
            "escalar a nivel 3",
            "urgencia alta reportada por el cliente"
        )
    
    elif tiempo_espera_minutos >= 60:
        return (
            "escalar a nivel 2",
            "tiempo de espera igual o superior a 60 minutos"
        )
    
    elif nivel_urgencia == "media" and tiempo_espera_minutos >= 30:
        return (
            "escalar a nivel 2",
            "urgencia media y tiempo de espera elevado"
        )
    
    elif cliente_premium and tiempo_espera_minutos >= 20:
        return (
            "escalar a nivel 2",
            "cliente premium con tiempo de espera elevado"
        )
    
    else:
        return (
            "mantener en nivel 1",
            "condiciones dentro de los límites normales"
        )
```

### Simulación y pruebas

Se probará el agente con diferentes combinaciones de tiempo de espera,
urgencia y condición premium.

Se incluyen situaciones normales y casos donde las tres condiciones no
coinciden, para verificar que el agente aplique correctamente las reglas.

```python
pruebas_soporte = [
    (5, "alta", False),
    (35, "media", False),
    (10, "baja", True),
    (25, "baja", True),
    (70, "baja", False),
    (15, "media", True),
    (45, "alta", True)
]

for tiempo, urgencia, premium in pruebas_soporte:
    
    accion, motivo = agente_soporte(
        tiempo,
        urgencia,
        premium
    )
    
    print(f"Tiempo de espera: {tiempo} minutos")
    print(f"Nivel de urgencia: {urgencia}")
    print(f"Cliente premium: {premium}")
    print(f"Acción: {accion}")
    print(f"Motivo: {motivo}")
    print("-" * 60)
```
